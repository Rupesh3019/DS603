```python
from collections import Counter
from functools import reduce

with open('Rupesh_date.txt', 'r', encoding='utf-8') as k:
    text = k.read()

words = text.split()

def mapper(word):
    return (word, 1)

def reducer(counts, w_c):
    word, count = w_c
    counts[word] += count
    return counts

no_words= reduce(reducer, map(mapper, words), Counter())

for word, count in no_words.items():
    print(' No. of times Repeated : '+ word+': '+ str(count))
```

```python
from spellchecker import SpellChecker
from collections import Counter
import re

sp = SpellChecker()

with open('Rupesh_year.txt' , 'r', encoding='utf-8') as r:
    text = r.read()
    
def word_pre_process(word):
    return re.sub(r'[^\w\s]', '', word)  # Removes all non-alphanumeric characters except spaces

words = text.split()

def mapper(word):
    word = word_pre_process(word)
    correction = sp.correction(word)
    if correction != word:
        return (word, 1)
    return (word, 0)

def reducer(counts, c_w):
    word, count = c_w
    counts[word] += count
    return counts

mis_spell= reduce(reducer, map(mapper, words), Counter())

for word, count in mis_spell.items():
    if count > 0:
        print(f'{word} : {count}')
```
